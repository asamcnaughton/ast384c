````python
import numpy as np
import scipy as sp 
import matplotlib.pyplot as plt
%matplotlib inline
np.random.seed(1)
````

Here's some pseudo code for the Metropolis algorithm (symmetric proposal --- $q(x|x') = q(x'|x)$

```
# initialize the current state x_0
x = x_0
samples[0] = x

for i in range(1, S): # S samples 
    x' ~ q(x'| x) # sample from the proposal
    
    r = min(1, p(x') / p(x)) # calculate acceptance rate

    # accept or reject
    u ~ U(0,1)
    if r > u: # accept
        x = x'
    else: 
        x = x
    samples[i] = x
```

Let's say we want to use MCMC to sample from a bimodal distribution

````python
_p0 = sp.stats.norm(loc=1, scale=6)
_p1 = sp.stats.norm(loc=-2, scale=1)
def p(x): 
    # this is your target distribution
    return 0.5*(_p0.pdf(x) + _p1.pdf(x))
````

Now, lets define a symmetric proposal distribution: a Gaussian

````python
def proposal_sample(x, scale=1): 
    # sample proposal distribution (simple Gaussian distribution)
    # x'~q(x'|x) = N(x, scale)
    q = sp.stats.norm(loc=x, scale=scale)
    return q.rvs()
````
